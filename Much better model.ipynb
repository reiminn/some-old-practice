```python
import pandas as pd
Data=pd.read_csv("House_price.csv")
#Data.head
Data.drop(["id","Lattitude","Longitude","Postal Code"],axis=1)
Data.isnull().sum()
```

```python
from sklearn.pipeline import Pipeline
from xgboost import XGBRegressor
Model=Pipeline(steps=[('model', XGBRegressor(n_estimators=1000 ,learning_rate=0.05))])
```

```python
from sklearn.model_selection import cross_val_score
y=Data.Price
X=Data.drop(['Price'],axis=1)
scores=-1*cross_val_score(Model,X,y,
                          cv=6,
                          scoring="neg_mean_absolute_error")
print("MAE:\n", scores)
print("Mean",scores.mean())
Model.fit(X,y)
```

```python
import joblib
```

```python
joblib.dump(Model, "Model_save")
```
